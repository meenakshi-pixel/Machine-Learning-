```python
# ============================================================
# HEART DISEASE — XGBOOST, LIGHTGBM & SHAP INTERPRETABILITY
# ============================================================

# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip install -q xgboost lightgbm shap

# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

import shap

# ============================================================
# 3. LOAD HEART DISEASE DATASET
# ============================================================

url = "https://raw.githubusercontent.com/plotly/datasets/master/heart.csv"

df = pd.read_csv(url)

print("First 5 rows:")
print(df.head())

# ============================================================
# 4. DATASET INFORMATION
# ============================================================

print("\n========================================")
print("DATASET INFORMATION")
print("========================================")

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# 5. BASIC STATISTICS
# ============================================================

print("\n========================================")
print("STATISTICAL SUMMARY")
print("========================================")

print(df.describe())

# ============================================================
# 6. CHECK TARGET DISTRIBUTION
# ============================================================

print("\n========================================")
print("TARGET DISTRIBUTION")
print("========================================")

print(df["target"].value_counts())

plt.figure(figsize=(6, 4))

df["target"].value_counts().sort_index().plot(
    kind="bar"
)

plt.xlabel("Heart Disease")
plt.ylabel("Number of Patients")

plt.title("Heart Disease Target Distribution")

plt.xticks(
    [0, 1],
    ["No Disease", "Disease"],
    rotation=0
)

plt.show()

# ============================================================
# 7. ENCODE CATEGORICAL VARIABLES
# ============================================================

# Convert categorical columns into numerical values
# using One-Hot Encoding.

categorical_columns = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]

df_encoded = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True
)

print("\n========================================")
print("DATA AFTER ENCODING")
print("========================================")

print(df_encoded.head())

print("\nNew Shape:")
print(df_encoded.shape)

# ============================================================
# 8. SEPARATE FEATURES AND TARGET
# ============================================================

X = df_encoded.drop(
    "target",
    axis=1
)

y = df_encoded["target"]

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print("target")

# ============================================================
# 9. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n========================================")
print("TRAIN TEST SPLIT")
print("========================================")

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])

# ============================================================
# 10. XGBOOST CLASSIFIER
# ============================================================

xgb_model = XGBClassifier(
    n_estimators=100,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="logloss"
)

xgb_model.fit(
    X_train,
    y_train
)

print("\nXGBoost model trained successfully.")

# ============================================================
# 11. XGBOOST PREDICTIONS
# ============================================================

xgb_pred = xgb_model.predict(X_test)

xgb_prob = xgb_model.predict_proba(
    X_test
)[:, 1]

# ============================================================
# 12. XGBOOST EVALUATION
# ============================================================

xgb_accuracy = accuracy_score(
    y_test,
    xgb_pred
)

xgb_precision = precision_score(
    y_test,
    xgb_pred
)

xgb_recall = recall_score(
    y_test,
    xgb_pred
)

xgb_f1 = f1_score(
    y_test,
    xgb_pred
)

xgb_auc = roc_auc_score(
    y_test,
    xgb_prob
)

print("\n========================================")
print("XGBOOST RESULTS")
print("========================================")

print("Accuracy :", xgb_accuracy)
print("Precision:", xgb_precision)
print("Recall   :", xgb_recall)
print("F1 Score :", xgb_f1)
print("ROC-AUC  :", xgb_auc)

# ============================================================
# 13. XGBOOST CLASSIFICATION REPORT
# ============================================================

print("\nXGBoost Classification Report:")

print(
    classification_report(
        y_test,
        xgb_pred,
        target_names=[
            "No Disease",
            "Disease"
        ]
    )
)

# ============================================================
# 14. XGBOOST CONFUSION MATRIX
# ============================================================

xgb_cm = confusion_matrix(
    y_test,
    xgb_pred
)

print("\nXGBoost Confusion Matrix:")
print(xgb_cm)

# ============================================================
# 15. LIGHTGBM CLASSIFIER
# ============================================================

lgbm_model = LGBMClassifier(
    n_estimators=100,
    max_depth=4,
    learning_rate=0.05,
    num_leaves=15,
    random_state=42,
    verbosity=-1
)

lgbm_model.fit(
    X_train,
    y_train
)

print("\nLightGBM model trained successfully.")

# ============================================================
# 16. LIGHTGBM PREDICTIONS
# ============================================================

lgbm_pred = lgbm_model.predict(
    X_test
)

lgbm_prob = lgbm_model.predict_proba(
    X_test
)[:, 1]

# ============================================================
# 17. LIGHTGBM EVALUATION
# ============================================================

lgbm_accuracy = accuracy_score(
    y_test,
    lgbm_pred
)

lgbm_precision = precision_score(
    y_test,
    lgbm_pred
)

lgbm_recall = recall_score(
    y_test,
    lgbm_pred
)

lgbm_f1 = f1_score(
    y_test,
    lgbm_pred
)

lgbm_auc = roc_auc_score(
    y_test,
    lgbm_prob
)

print("\n========================================")
print("LIGHTGBM RESULTS")
print("========================================")

print("Accuracy :", lgbm_accuracy)
print("Precision:", lgbm_precision)
print("Recall   :", lgbm_recall)
print("F1 Score :", lgbm_f1)
print("ROC-AUC  :", lgbm_auc)

# ============================================================
# 18. LIGHTGBM CLASSIFICATION REPORT
# ============================================================

print("\nLightGBM Classification Report:")

print(
    classification_report(
        y_test,
        lgbm_pred,
        target_names=[
            "No Disease",
            "Disease"
        ]
    )
)

# ============================================================
# 19. LIGHTGBM CONFUSION MATRIX
# ============================================================

lgbm_cm = confusion_matrix(
    y_test,
    lgbm_pred
)

print("\nLightGBM Confusion Matrix:")
print(lgbm_cm)

# ============================================================
# 20. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame({

    "Model": [
        "XGBoost",
        "LightGBM"
    ],

    "Accuracy": [
        xgb_accuracy,
        lgbm_accuracy
    ],

    "Precision": [
        xgb_precision,
        lgbm_precision
    ],

    "Recall": [
        xgb_recall,
        lgbm_recall
    ],

    "F1 Score": [
        xgb_f1,
        lgbm_f1
    ],

    "ROC-AUC": [
        xgb_auc,
        lgbm_auc
    ]
})

print("\n========================================")
print("MODEL COMPARISON")
print("========================================")

print(comparison)

# ============================================================
# 21. MODEL COMPARISON GRAPH
# ============================================================

comparison.set_index(
    "Model"
).plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title(
    "XGBoost vs LightGBM Performance"
)

plt.ylabel("Score")

plt.ylim(0, 1)

plt.xticks(
    rotation=0
)

plt.show()

# ============================================================
# 22. XGBOOST FEATURE IMPORTANCE
# ============================================================

xgb_importance = pd.DataFrame({

    "Feature": X.columns,

    "Importance": xgb_model.feature_importances_

})

xgb_importance = xgb_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\n========================================")
print("XGBOOST FEATURE IMPORTANCE")
print("========================================")

print(xgb_importance)

# ============================================================
# 23. XGBOOST FEATURE IMPORTANCE GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    xgb_importance["Feature"],
    xgb_importance["Importance"]
)

plt.xlabel("Features")
plt.ylabel("Importance")

plt.title(
    "XGBoost Feature Importance"
)

plt.xticks(
    rotation=45
)

plt.show()

# ============================================================
# 24. LIGHTGBM FEATURE IMPORTANCE
# ============================================================

lgbm_importance = pd.DataFrame({

    "Feature": X.columns,

    "Importance": lgbm_model.feature_importances_

})

lgbm_importance = lgbm_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\n========================================")
print("LIGHTGBM FEATURE IMPORTANCE")
print("========================================")

print(lgbm_importance)

# ============================================================
# 25. LIGHTGBM FEATURE IMPORTANCE GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

plt.bar(
    lgbm_importance["Feature"],
    lgbm_importance["Importance"]
)

plt.xlabel("Features")
plt.ylabel("Importance")

plt.title(
    "LightGBM Feature Importance"
)

plt.xticks(
    rotation=45
)

plt.show()

# ============================================================
# 26. SHAP EXPLAINER FOR XGBOOST
# ============================================================

print("\n========================================")
print("SHAP INTERPRETABILITY - XGBOOST")
print("========================================")

explainer_xgb = shap.TreeExplainer(
    xgb_model
)

shap_values_xgb = explainer_xgb.shap_values(
    X_test
)

print("SHAP values calculated for XGBoost.")

# ============================================================
# 27. XGBOOST SHAP SUMMARY PLOT
# ============================================================

shap.summary_plot(
    shap_values_xgb,
    X_test
)

# ============================================================
# 28. XGBOOST SHAP BAR PLOT
# ============================================================

shap.summary_plot(
    shap_values_xgb,
    X_test,
    plot_type="bar"
)

# ============================================================
# 29. SHAP EXPLAINER FOR LIGHTGBM
# ============================================================

print("\n========================================")
print("SHAP INTERPRETABILITY - LIGHTGBM")
print("========================================")

explainer_lgbm = shap.TreeExplainer(
    lgbm_model
)

shap_values_lgbm = explainer_lgbm.shap_values(
    X_test
)

print("SHAP values calculated for LightGBM.")

# ============================================================
# 30. LIGHTGBM SHAP SUMMARY PLOT
# ============================================================

# For binary classification, SHAP may return
# a list containing values for both classes.

if isinstance(shap_values_lgbm, list):

    shap.summary_plot(
        shap_values_lgbm[1],
        X_test
    )

else:

    shap.summary_plot(
        shap_values_lgbm,
        X_test
    )

# ============================================================
# 31. LIGHTGBM SHAP BAR PLOT
# ============================================================

if isinstance(shap_values_lgbm, list):

    shap.summary_plot(
        shap_values_lgbm[1],
        X_test,
        plot_type="bar"
    )

else:

    shap.summary_plot(
        shap_values_lgbm,
        X_test,
        plot_type="bar"
    )

# ============================================================
# 32. SHAP WATERFALL PLOT FOR ONE PATIENT
# ============================================================

# Explain one test sample

sample_index = 0

if isinstance(shap_values_xgb, list):

    sample_shap_values = shap_values_xgb[1]

else:

    sample_shap_values = shap_values_xgb

# Create SHAP Explanation object

explanation = shap.Explanation(
    values=sample_shap_values[sample_index],
    base_values=explainer_xgb.expected_value,
    data=X_test.iloc[sample_index],
    feature_names=X_test.columns
)

shap.plots.waterfall(
    explanation
)

# ============================================================
# 33. SELECT BEST MODEL
# ============================================================

if xgb_accuracy > lgbm_accuracy:

    best_model = "XGBoost"
    best_accuracy = xgb_accuracy

else:

    best_model = "LightGBM"
    best_accuracy = lgbm_accuracy

# ============================================================
# 34. FINAL RESULT
# ============================================================

print("\n========================================")
print("FINAL RESULT")
print("========================================")

print("Dataset       : Heart Disease")
print("Task          : Binary Classification")
print()
print("Models Used:")
print("1. XGBoost")
print("2. LightGBM")
print("3. SHAP Interpretability")
print()
print("XGBoost Accuracy :", xgb_accuracy)
print("LightGBM Accuracy:", lgbm_accuracy)
print()
print("Best Model:", best_model)
print("Best Accuracy:", best_accuracy)

print("\nProject Completed Successfully!")
```
