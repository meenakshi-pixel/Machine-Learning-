```python
# ============================================================
# PROJECT 4: CALIFORNIA HOUSING
# LINEAR REGRESSION AND REGULARISATION
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

housing = fetch_california_housing()

X = pd.DataFrame(
    housing.data,
    columns=housing.feature_names
)

y = pd.Series(
    housing.target,
    name="MedHouseVal"
)

print("Dataset loaded successfully!")

# ------------------------------------------------------------
# 3. DISPLAY DATA
# ------------------------------------------------------------

print("\nFirst 5 rows:")
display(X.head())

print("\nTarget values:")
display(y.head())

# ------------------------------------------------------------
# 4. DATASET SHAPE
# ------------------------------------------------------------

print("\nFeature Shape:")
print(X.shape)

print("\nTarget Shape:")
print(y.shape)

# ------------------------------------------------------------
# 5. FEATURE NAMES
# ------------------------------------------------------------

print("\nFeature Names:")
print(X.columns.tolist())

# ------------------------------------------------------------
# 6. COMPLETE DATASET
# ------------------------------------------------------------

df = X.copy()

df["MedHouseVal"] = y

print("\nComplete Dataset:")
display(df.head())

# ------------------------------------------------------------
# 7. DATASET INFORMATION
# ------------------------------------------------------------

print("\nDataset Information:")
df.info()

# ------------------------------------------------------------
# 8. STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\nStatistical Summary:")
display(df.describe())

# ------------------------------------------------------------
# 9. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 10. CHECK DUPLICATES
# ------------------------------------------------------------

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 11. REMOVE DUPLICATES
# ------------------------------------------------------------

df = df.drop_duplicates()

print("\nShape After Removing Duplicates:")
print(df.shape)

# ------------------------------------------------------------
# 12. TARGET DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    y,
    bins=40
)

plt.title("Distribution of Median House Value")
plt.xlabel("Median House Value")
plt.ylabel("Frequency")

plt.show()

# ------------------------------------------------------------
# 13. FEATURE DISTRIBUTIONS
# ------------------------------------------------------------

for column in X.columns:

    plt.figure(figsize=(7, 4))

    plt.hist(
        X[column],
        bins=30
    )

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.show()

# ------------------------------------------------------------
# 14. CORRELATION MATRIX
# ------------------------------------------------------------

correlation_matrix = df.corr()

plt.figure(figsize=(12, 8))

plt.imshow(
    correlation_matrix,
    interpolation="nearest"
)

plt.colorbar()

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Matrix")

plt.tight_layout()

plt.show()

# ------------------------------------------------------------
# 15. CORRELATION WITH TARGET
# ------------------------------------------------------------

target_correlation = (
    df.corr()["MedHouseVal"]
    .sort_values(ascending=False)
)

print("\nCorrelation with Target:")
print(target_correlation)

# ------------------------------------------------------------
# 16. FEATURE VS TARGET
# ------------------------------------------------------------

for column in X.columns:

    plt.figure(figsize=(7, 5))

    plt.scatter(
        X[column],
        y,
        alpha=0.3
    )

    plt.title(f"{column} vs Median House Value")
    plt.xlabel(column)
    plt.ylabel("Median House Value")

    plt.show()

# ============================================================
# MACHINE LEARNING
# ============================================================

# ------------------------------------------------------------
# 17. DEFINE FEATURES AND TARGET
# ------------------------------------------------------------

X = df.drop(
    "MedHouseVal",
    axis=1
)

y = df["MedHouseVal"]

print("\nFeatures:")
display(X.head())

print("\nTarget:")
display(y.head())

# ------------------------------------------------------------
# 18. TRAIN-TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Data Shape:")
print(X_train.shape)

print("\nTesting Data Shape:")
print(X_test.shape)

# ------------------------------------------------------------
# 19. FEATURE SCALING
# ------------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("\nFeature scaling completed.")

# ============================================================
# LINEAR REGRESSION
# ============================================================

# ------------------------------------------------------------
# 20. CREATE LINEAR REGRESSION MODEL
# ------------------------------------------------------------

linear_model = LinearRegression()

# ------------------------------------------------------------
# 21. TRAIN MODEL
# ------------------------------------------------------------

linear_model.fit(
    X_train_scaled,
    y_train
)

print("\nLinear Regression model trained.")

# ------------------------------------------------------------
# 22. PREDICTIONS
# ------------------------------------------------------------

y_pred_linear = linear_model.predict(
    X_test_scaled
)

# ------------------------------------------------------------
# 23. LINEAR REGRESSION EVALUATION
# ------------------------------------------------------------

linear_mae = mean_absolute_error(
    y_test,
    y_pred_linear
)

linear_mse = mean_squared_error(
    y_test,
    y_pred_linear
)

linear_rmse = np.sqrt(
    linear_mse
)

linear_r2 = r2_score(
    y_test,
    y_pred_linear
)

print("\n==============================")
print("LINEAR REGRESSION RESULTS")
print("==============================")

print("MAE :", linear_mae)
print("MSE :", linear_mse)
print("RMSE:", linear_rmse)
print("R²  :", linear_r2)

# ------------------------------------------------------------
# 24. ACTUAL VS PREDICTED
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

plt.scatter(
    y_test,
    y_pred_linear,
    alpha=0.4
)

plt.xlabel("Actual House Value")
plt.ylabel("Predicted House Value")

plt.title(
    "Linear Regression: Actual vs Predicted"
)

plt.show()

# ============================================================
# RIDGE REGRESSION
# ============================================================

# ------------------------------------------------------------
# 25. RIDGE REGRESSION
# ------------------------------------------------------------

ridge_model = Ridge(
    alpha=1.0
)

# ------------------------------------------------------------
# 26. TRAIN RIDGE MODEL
# ------------------------------------------------------------

ridge_model.fit(
    X_train_scaled,
    y_train
)

print("\nRidge Regression model trained.")

# ------------------------------------------------------------
# 27. RIDGE PREDICTION
# ------------------------------------------------------------

y_pred_ridge = ridge_model.predict(
    X_test_scaled
)

# ------------------------------------------------------------
# 28. RIDGE EVALUATION
# ------------------------------------------------------------

ridge_mae = mean_absolute_error(
    y_test,
    y_pred_ridge
)

ridge_mse = mean_squared_error(
    y_test,
    y_pred_ridge
)

ridge_rmse = np.sqrt(
    ridge_mse
)

ridge_r2 = r2_score(
    y_test,
    y_pred_ridge
)

print("\n==============================")
print("RIDGE REGRESSION RESULTS")
print("==============================")

print("MAE :", ridge_mae)
print("MSE :", ridge_mse)
print("RMSE:", ridge_rmse)
print("R²  :", ridge_r2)

# ============================================================
# LASSO REGRESSION
# ============================================================

# ------------------------------------------------------------
# 29. LASSO REGRESSION
# ------------------------------------------------------------

lasso_model = Lasso(
    alpha=0.01,
    max_iter=10000
)

# ------------------------------------------------------------
# 30. TRAIN LASSO MODEL
# ------------------------------------------------------------

lasso_model.fit(
    X_train_scaled,
    y_train
)

print("\nLasso Regression model trained.")

# ------------------------------------------------------------
# 31. LASSO PREDICTION
# ------------------------------------------------------------

y_pred_lasso = lasso_model.predict(
    X_test_scaled
)

# ------------------------------------------------------------
# 32. LASSO EVALUATION
# ------------------------------------------------------------

lasso_mae = mean_absolute_error(
    y_test,
    y_pred_lasso
)

lasso_mse = mean_squared_error(
    y_test,
    y_pred_lasso
)

lasso_rmse = np.sqrt(
    lasso_mse
)

lasso_r2 = r2_score(
    y_test,
    y_pred_lasso
)

print("\n==============================")
print("LASSO REGRESSION RESULTS")
print("==============================")

print("MAE :", lasso_mae)
print("MSE :", lasso_mse)
print("RMSE:", lasso_rmse)
print("R²  :", lasso_r2)

# ============================================================
# MODEL COMPARISON
# ============================================================

# ------------------------------------------------------------
# 33. CREATE COMPARISON TABLE
# ------------------------------------------------------------

results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Ridge Regression",
        "Lasso Regression"
    ],

    "MAE": [
        linear_mae,
        ridge_mae,
        lasso_mae
    ],

    "MSE": [
        linear_mse,
        ridge_mse,
        lasso_mse
    ],

    "RMSE": [
        linear_rmse,
        ridge_rmse,
        lasso_rmse
    ],

    "R2 Score": [
        linear_r2,
        ridge_r2,
        lasso_r2
    ]
})

print("\n==============================")
print("MODEL COMPARISON")
print("==============================")

display(results)

# ------------------------------------------------------------
# 34. R2 SCORE COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["R2 Score"]
)

plt.title("R² Score Comparison")
plt.xlabel("Model")
plt.ylabel("R² Score")

plt.xticks(rotation=20)

plt.show()

# ------------------------------------------------------------
# 35. RMSE COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    results["Model"],
    results["RMSE"]
)

plt.title("RMSE Comparison")
plt.xlabel("Model")
plt.ylabel("RMSE")

plt.xticks(rotation=20)

plt.show()

# ============================================================
# RIDGE ALPHA EXPERIMENT
# ============================================================

# ------------------------------------------------------------
# 36. TEST DIFFERENT RIDGE ALPHA VALUES
# ------------------------------------------------------------

alphas = [
    0.01,
    0.1,
    1,
    10,
    100
]

ridge_scores = []

for alpha in alphas:

    model = Ridge(
        alpha=alpha
    )

    model.fit(
        X_train_scaled,
        y_train
    )

    predictions = model.predict(
        X_test_scaled
    )

    score = r2_score(
        y_test,
        predictions
    )

    ridge_scores.append(score)

# ------------------------------------------------------------
# 37. DISPLAY RIDGE ALPHA RESULTS
# ------------------------------------------------------------

ridge_alpha_results = pd.DataFrame({
    "Alpha": alphas,
    "R2 Score": ridge_scores
})

print("\nRidge Alpha Results:")
display(ridge_alpha_results)

# ------------------------------------------------------------
# 38. RIDGE ALPHA PLOT
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    alphas,
    ridge_scores,
    marker="o"
)

plt.xscale("log")

plt.title("Ridge Alpha vs R² Score")
plt.xlabel("Alpha")
plt.ylabel("R² Score")

plt.show()

# ============================================================
# LASSO ALPHA EXPERIMENT
# ============================================================

# ------------------------------------------------------------
# 39. TEST DIFFERENT LASSO ALPHA VALUES
# ------------------------------------------------------------

lasso_alphas = [
    0.0001,
    0.001,
    0.01,
    0.1,
    1
]

lasso_scores = []

for alpha in lasso_alphas:

    model = Lasso(
        alpha=alpha,
        max_iter=10000
    )

    model.fit(
        X_train_scaled,
        y_train
    )

    predictions = model.predict(
        X_test_scaled
    )

    score = r2_score(
        y_test,
        predictions
    )

    lasso_scores.append(score)

# ------------------------------------------------------------
# 40. DISPLAY LASSO RESULTS
# ------------------------------------------------------------

lasso_alpha_results = pd.DataFrame({
    "Alpha": lasso_alphas,
    "R2 Score": lasso_scores
})

print("\nLasso Alpha Results:")
display(lasso_alpha_results)

# ------------------------------------------------------------
# 41. LASSO ALPHA PLOT
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    lasso_alphas,
    lasso_scores,
    marker="o"
)

plt.xscale("log")

plt.title("Lasso Alpha vs R² Score")
plt.xlabel("Alpha")
plt.ylabel("R² Score")

plt.show()

# ============================================================
# FEATURE COEFFICIENTS
# ============================================================

# ------------------------------------------------------------
# 42. LINEAR REGRESSION COEFFICIENTS
# ------------------------------------------------------------

linear_coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": linear_model.coef_
})

print("\nLinear Regression Coefficients:")
display(
    linear_coefficients.sort_values(
        by="Coefficient",
        ascending=False
    )
)

# ------------------------------------------------------------
# 43. RIDGE COEFFICIENTS
# ------------------------------------------------------------

ridge_coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": ridge_model.coef_
})

print("\nRidge Regression Coefficients:")
display(
    ridge_coefficients.sort_values(
        by="Coefficient",
        ascending=False
    )
)

# ------------------------------------------------------------
# 44. LASSO COEFFICIENTS
# ------------------------------------------------------------

lasso_coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": lasso_model.coef_
})

print("\nLasso Regression Coefficients:")
display(
    lasso_coefficients.sort_values(
        by="Coefficient",
        ascending=False
    )
)

# ------------------------------------------------------------
# 45. CHECK ZERO COEFFICIENTS IN LASSO
# ------------------------------------------------------------

zero_coefficients = (
    lasso_model.coef_ == 0
)

print(
    "\nNumber of features with zero Lasso coefficients:"
)

print(
    zero_coefficients.sum()
)

# ============================================================
# FINAL CONCLUSION
# ============================================================

print("""
============================================================
CONCLUSION
============================================================

1. California Housing data was explored using EDA.

2. Missing values and duplicate rows were checked.

3. Features were standardized using StandardScaler.

4. Linear Regression was used as the baseline model.

5. Ridge Regression was implemented to reduce the effect
   of large coefficients using L2 regularisation.

6. Lasso Regression was implemented using L1 regularisation.

7. MAE, MSE, RMSE and R² were used for model evaluation.

8. Different alpha values were tested for Ridge and Lasso.

9. Lasso can perform feature selection by shrinking some
   coefficients to zero.

10. The models were compared using their evaluation metrics.
============================================================
""")
```
