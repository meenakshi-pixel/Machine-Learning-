```python
# ============================================================
# PROJECT 9: HEART DISEASE
# RANDOM FOREST ENSEMBLE
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

from google.colab import files

uploaded = files.upload()

# Change filename if necessary
df = pd.read_csv("heart.csv")

print("Dataset loaded successfully!")

# ------------------------------------------------------------
# 3. BASIC DATA EXPLORATION
# ------------------------------------------------------------

print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

# ------------------------------------------------------------
# 4. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 5. CHECK DUPLICATES
# ------------------------------------------------------------

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 6. REMOVE DUPLICATES
# ------------------------------------------------------------

df = df.drop_duplicates()

print("\nShape After Removing Duplicates:")
print(df.shape)

# ============================================================
# EXPLORATORY DATA ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 7. TARGET DISTRIBUTION
# ------------------------------------------------------------

print("\nTarget Distribution:")
print(df["target"].value_counts())

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="target"
)

plt.title("Heart Disease Target Distribution")
plt.xlabel("Target")
plt.ylabel("Count")

plt.show()

# ------------------------------------------------------------
# 8. TARGET PERCENTAGE
# ------------------------------------------------------------

print("\nTarget Percentage:")

target_percentage = (
    df["target"]
    .value_counts(normalize=True)
    * 100
)

print(target_percentage)

# ------------------------------------------------------------
# 9. AGE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="age",
    bins=20,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()

# ------------------------------------------------------------
# 10. AGE VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="target",
    y="age"
)

plt.title("Age vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Age")

plt.show()

# ------------------------------------------------------------
# 11. CHEST PAIN VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="cp",
    hue="target"
)

plt.title("Chest Pain Type vs Heart Disease")
plt.xlabel("Chest Pain Type")
plt.ylabel("Count")

plt.show()

# ------------------------------------------------------------
# 12. MAXIMUM HEART RATE VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="target",
    y="thalach"
)

plt.title("Maximum Heart Rate vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Maximum Heart Rate")

plt.show()

# ------------------------------------------------------------
# 13. CHOLESTEROL VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="target",
    y="chol"
)

plt.title("Cholesterol vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Cholesterol")

plt.show()

# ------------------------------------------------------------
# 14. CORRELATION MATRIX
# ------------------------------------------------------------

plt.figure(figsize=(12, 9))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Heart Disease Correlation Matrix")

plt.show()

# ------------------------------------------------------------
# 15. CORRELATION WITH TARGET
# ------------------------------------------------------------

target_correlation = (
    df.corr(numeric_only=True)["target"]
    .sort_values(ascending=False)
)

print("\nCorrelation with Target:")
print(target_correlation)

# ============================================================
# DATA PREPROCESSING
# ============================================================

# ------------------------------------------------------------
# 16. DEFINE FEATURES AND TARGET
# ------------------------------------------------------------

X = df.drop(
    columns=["target"]
)

y = df["target"]

print("\nFeatures:")
display(X.head())

print("\nTarget:")
display(y.head())

# ------------------------------------------------------------
# 17. CHECK TARGET CLASSES
# ------------------------------------------------------------

print("\nTarget Classes:")
print(sorted(y.unique()))

print("\nNumber of Classes:")
print(y.nunique())

# ------------------------------------------------------------
# 18. TRAIN-TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Shape:")
print(X_train.shape)

print("\nTesting Shape:")
print(X_test.shape)

# ============================================================
# RANDOM FOREST
# ============================================================

# ------------------------------------------------------------
# 19. CREATE RANDOM FOREST MODEL
# ------------------------------------------------------------

rf_model = RandomForestClassifier(
    n_estimators=100,
    criterion="gini",
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    bootstrap=True,
    random_state=42,
    n_jobs=-1
)

# ------------------------------------------------------------
# 20. TRAIN RANDOM FOREST
# ------------------------------------------------------------

rf_model.fit(
    X_train,
    y_train
)

print("\nRandom Forest model trained successfully.")

# ------------------------------------------------------------
# 21. PREDICTIONS
# ------------------------------------------------------------

y_pred = rf_model.predict(
    X_test
)

# ------------------------------------------------------------
# 22. PREDICT PROBABILITIES
# ------------------------------------------------------------

y_probability = rf_model.predict_proba(
    X_test
)

print("\nFirst 10 Predictions:")
print(y_pred[:10])

# ============================================================
# MODEL EVALUATION
# ============================================================

# ------------------------------------------------------------
# 23. ACCURACY
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_test,
    y_pred
)

# ------------------------------------------------------------
# 24. PRECISION
# ------------------------------------------------------------

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

# ------------------------------------------------------------
# 25. RECALL
# ------------------------------------------------------------

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

# ------------------------------------------------------------
# 26. F1 SCORE
# ------------------------------------------------------------

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\n==============================")
print("RANDOM FOREST RESULTS")
print("==============================")

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")

# ------------------------------------------------------------
# 27. CLASSIFICATION REPORT
# ------------------------------------------------------------

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)

# ============================================================
# CONFUSION MATRIX
# ============================================================

# ------------------------------------------------------------
# 28. CONFUSION MATRIX
# ------------------------------------------------------------

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)

# ------------------------------------------------------------
# 29. CONFUSION MATRIX VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=rf_model.classes_,
    yticklabels=rf_model.classes_
)

plt.title(
    "Random Forest Confusion Matrix"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()

# ============================================================
# FEATURE IMPORTANCE
# ============================================================

# ------------------------------------------------------------
# 30. FEATURE IMPORTANCE
# ------------------------------------------------------------

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nRandom Forest Feature Importance:")

display(feature_importance)

# ------------------------------------------------------------
# 31. FEATURE IMPORTANCE PLOT
# ------------------------------------------------------------

plt.figure(figsize=(10, 7))

plt.barh(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.title(
    "Random Forest Feature Importance"
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.gca().invert_yaxis()

plt.show()

# ============================================================
# NUMBER OF TREES EXPERIMENT
# ============================================================

# ------------------------------------------------------------
# 32. TEST DIFFERENT NUMBER OF TREES
# ------------------------------------------------------------

n_estimators_values = [
    10,
    25,
    50,
    100,
    150,
    200,
    300
]

tree_scores = []

for n in n_estimators_values:

    model = RandomForestClassifier(
        n_estimators=n,
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train,
        y_train
    )

    prediction = model.predict(
        X_test
    )

    score = accuracy_score(
        y_test,
        prediction
    )

    tree_scores.append(
        score
    )

# ------------------------------------------------------------
# 33. TREE COUNT RESULTS
# ------------------------------------------------------------

tree_results = pd.DataFrame({
    "Number of Trees": n_estimators_values,
    "Test Accuracy": tree_scores
})

print("\nNumber of Trees Experiment:")

display(tree_results)

# ------------------------------------------------------------
# 34. TREE COUNT PLOT
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.plot(
    n_estimators_values,
    tree_scores,
    marker="o"
)

plt.title(
    "Number of Trees vs Accuracy"
)

plt.xlabel("Number of Trees")
plt.ylabel("Test Accuracy")

plt.show()

# ============================================================
# MAX DEPTH EXPERIMENT
# ============================================================

# ------------------------------------------------------------
# 35. TEST DIFFERENT MAX DEPTHS
# ------------------------------------------------------------

depth_values = [
    2,
    3,
    4,
    5,
    6,
    8,
    10,
    15,
    None
]

depth_scores = []

for depth in depth_values:

    model = RandomForestClassifier(
        n_estimators=100,
        max_depth=depth,
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train,
        y_train
    )

    prediction = model.predict(
        X_test
    )

    score = accuracy_score(
        y_test,
        prediction
    )

    depth_scores.append(
        score
    )

# ------------------------------------------------------------
# 36. DEPTH RESULTS
# ------------------------------------------------------------

depth_results = pd.DataFrame({
    "Max Depth": [
        str(depth)
        for depth in depth_values
    ],
    "Test Accuracy": depth_scores
})

print("\nMaximum Depth Experiment:")

display(depth_results)

# ------------------------------------------------------------
# 37. DEPTH PLOT
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.plot(
    range(len(depth_values)),
    depth_scores,
    marker="o"
)

plt.xticks(
    range(len(depth_values)),
    [
        str(depth)
        for depth in depth_values
    ]
)

plt.title(
    "Maximum Depth vs Accuracy"
)

plt.xlabel("Maximum Depth")
plt.ylabel("Test Accuracy")

plt.show()

# ============================================================
# FINAL RANDOM FOREST MODEL
# ============================================================

# ------------------------------------------------------------
# 38. BEST NUMBER OF TREES
# ------------------------------------------------------------

best_tree_index = np.argmax(
    tree_scores
)

best_n_estimators = n_estimators_values[
    best_tree_index
]

print(
    "\nBest Number of Trees:",
    best_n_estimators
)

# ------------------------------------------------------------
# 39. CREATE FINAL MODEL
# ------------------------------------------------------------

final_model = RandomForestClassifier(
    n_estimators=best_n_estimators,
    criterion="gini",
    max_features="sqrt",
    random_state=42,
    n_jobs=-1
)

# ------------------------------------------------------------
# 40. TRAIN FINAL MODEL
# ------------------------------------------------------------

final_model.fit(
    X_train,
    y_train
)

# ------------------------------------------------------------
# 41. FINAL PREDICTIONS
# ------------------------------------------------------------

final_prediction = final_model.predict(
    X_test
)

# ------------------------------------------------------------
# 42. FINAL ACCURACY
# ------------------------------------------------------------

final_accuracy = accuracy_score(
    y_test,
    final_prediction
)

final_precision = precision_score(
    y_test,
    final_prediction,
    average="weighted",
    zero_division=0
)

final_recall = recall_score(
    y_test,
    final_prediction,
    average="weighted",
    zero_division=0
)

final_f1 = f1_score(
    y_test,
    final_prediction,
    average="weighted",
    zero_division=0
)

print("\n==============================")
print("FINAL RANDOM FOREST MODEL")
print("==============================")

print(f"Number of Trees: {best_n_estimators}")
print(f"Accuracy       : {final_accuracy:.4f}")
print(f"Precision      : {final_precision:.4f}")
print(f"Recall         : {final_recall:.4f}")
print(f"F1 Score       : {final_f1:.4f}")

# ------------------------------------------------------------
# 43. FINAL CLASSIFICATION REPORT
# ------------------------------------------------------------

print("\nFinal Classification Report:")

print(
    classification_report(
        y_test,
        final_prediction,
        zero_division=0
    )
)

# ------------------------------------------------------------
# 44. FINAL CONFUSION MATRIX
# ------------------------------------------------------------

final_cm = confusion_matrix(
    y_test,
    final_prediction
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    final_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=final_model.classes_,
    yticklabels=final_model.classes_
)

plt.title(
    "Final Random Forest Confusion Matrix"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()

# ============================================================
# ACTUAL VS PREDICTED
# ============================================================

# ------------------------------------------------------------
# 45. PREDICTION TABLE
# ------------------------------------------------------------

prediction_results = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": final_prediction
})

prediction_results["Correct"] = (
    prediction_results["Actual"]
    ==
    prediction_results["Predicted"]
)

print("\nActual vs Predicted:")

display(
    prediction_results.head(20)
)

# ============================================================
# SAMPLE PATIENT PREDICTION
# ============================================================

# ------------------------------------------------------------
# 46. SELECT SAMPLE
# ------------------------------------------------------------

sample = X_test.iloc[[0]]

print("\nSample Patient:")

display(sample)

# ------------------------------------------------------------
# 47. PREDICT SAMPLE
# ------------------------------------------------------------

sample_prediction = final_model.predict(
    sample
)[0]

sample_probability = final_model.predict_proba(
    sample
)[0]

print("\n==============================")
print("SAMPLE PATIENT PREDICTION")
print("==============================")

print(
    "Predicted Heart Disease Class:",
    sample_prediction
)

print("\nPrediction Probabilities:")

for class_label, probability in zip(
    final_model.classes_,
    sample_probability
):

    print(
        f"Class {class_label}: "
        f"{probability:.2%}"
    )

# ============================================================
# MODEL COMPARISON WITH DECISION TREE
# ============================================================

# ------------------------------------------------------------
# 48. TRAIN SIMPLE DECISION TREE FOR COMPARISON
# ------------------------------------------------------------

from sklearn.tree import DecisionTreeClassifier

decision_tree = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

decision_tree.fit(
    X_train,
    y_train
)

dt_prediction = decision_tree.predict(
    X_test
)

dt_accuracy = accuracy_score(
    y_test,
    dt_prediction
)

# ------------------------------------------------------------
# 49. MODEL COMPARISON
# ------------------------------------------------------------

comparison = pd.DataFrame({

    "Model": [
        "Decision Tree",
        "Random Forest"
    ],

    "Accuracy": [
        dt_accuracy,
        final_accuracy
    ],

    "Precision": [
        precision_score(
            y_test,
            dt_prediction,
            average="weighted",
            zero_division=0
        ),
        final_precision
    ],

    "Recall": [
        recall_score(
            y_test,
            dt_prediction,
            average="weighted",
            zero_division=0
        ),
        final_recall
    ],

    "F1 Score": [
        f1_score(
            y_test,
            dt_prediction,
            average="weighted",
            zero_division=0
        ),
        final_f1
    ]
})

print("\n==============================")
print("DECISION TREE VS RANDOM FOREST")
print("==============================")

display(comparison)

# ------------------------------------------------------------
# 50. MODEL ACCURACY COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    comparison["Model"],
    comparison["Accuracy"]
)

plt.title(
    "Decision Tree vs Random Forest Accuracy"
)

plt.xlabel("Model")
plt.ylabel("Accuracy")

plt.show()

# ============================================================
# FINAL CONCLUSION
# ============================================================

print("""
============================================================
CONCLUSION
============================================================

1. The Heart Disease dataset was loaded and explored.

2. Missing values and duplicate records were checked.

3. EDA was performed using distributions, boxplots and
   correlation analysis.

4. The dataset was divided into training and testing sets.

5. A Random Forest classifier was trained using multiple
   decision trees.

6. Model performance was evaluated using:
   - Accuracy
   - Precision
   - Recall
   - F1 Score
   - Confusion Matrix

7. Feature importance was analyzed to identify the most
   influential features.

8. Different numbers of trees were tested.

9. Different maximum tree depths were tested.

10. A final Random Forest model was created.

11. A sample patient's heart disease class was predicted.

12. Random Forest was compared with a single Decision Tree
    to demonstrate the benefit of ensemble learning.

============================================================
""")
```
