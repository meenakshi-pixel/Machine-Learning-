```python
# ============================================================
# PROJECT 8: HEART DISEASE
# DECISION TREE CLASSIFICATION
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split

from sklearn.tree import (
    DecisionTreeClassifier,
    plot_tree
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

from google.colab import files

uploaded = files.upload()

# Change filename if necessary
df = pd.read_csv("heart.csv")

print("Dataset loaded successfully!")

# ------------------------------------------------------------
# 3. BASIC DATA EXPLORATION
# ------------------------------------------------------------

print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

# ------------------------------------------------------------
# 4. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 5. CHECK DUPLICATES
# ------------------------------------------------------------

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 6. REMOVE DUPLICATES
# ------------------------------------------------------------

df = df.drop_duplicates()

print("\nShape After Removing Duplicates:")
print(df.shape)

# ============================================================
# TARGET ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 7. CHECK TARGET COLUMN
# ------------------------------------------------------------

print("\nTarget Distribution:")

print(
    df["target"].value_counts()
)

# ------------------------------------------------------------
# 8. TARGET VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="target"
)

plt.title("Heart Disease Target Distribution")
plt.xlabel("Target")
plt.ylabel("Count")

plt.show()

# ------------------------------------------------------------
# 9. TARGET PERCENTAGE
# ------------------------------------------------------------

target_percentage = (
    df["target"]
    .value_counts(normalize=True)
    * 100
)

print("\nTarget Percentage:")

print(target_percentage)

# ============================================================
# EXPLORATORY DATA ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 10. AGE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="age",
    bins=20,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()

# ------------------------------------------------------------
# 11. AGE VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x="target",
    y="age"
)

plt.title("Age vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Age")

plt.show()

# ------------------------------------------------------------
# 12. SEX VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="sex",
    hue="target"
)

plt.title("Sex vs Heart Disease")
plt.xlabel("Sex")
plt.ylabel("Count")

plt.show()

# ------------------------------------------------------------
# 13. CHEST PAIN TYPE VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="cp",
    hue="target"
)

plt.title("Chest Pain Type vs Heart Disease")
plt.xlabel("Chest Pain Type")
plt.ylabel("Count")

plt.show()

# ------------------------------------------------------------
# 14. MAXIMUM HEART RATE VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="target",
    y="thalach"
)

plt.title("Maximum Heart Rate vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Maximum Heart Rate")

plt.show()

# ------------------------------------------------------------
# 15. CHOLESTEROL VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="target",
    y="chol"
)

plt.title("Cholesterol vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Cholesterol")

plt.show()

# ------------------------------------------------------------
# 16. RESTING BLOOD PRESSURE VS TARGET
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="target",
    y="trestbps"
)

plt.title("Resting Blood Pressure vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Resting Blood Pressure")

plt.show()

# ------------------------------------------------------------
# 17. CORRELATION MATRIX
# ------------------------------------------------------------

plt.figure(figsize=(12, 9))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Heart Disease Correlation Matrix")

plt.show()

# ------------------------------------------------------------
# 18. CORRELATION WITH TARGET
# ------------------------------------------------------------

target_correlation = (
    df.corr(numeric_only=True)["target"]
    .sort_values(ascending=False)
)

print("\nCorrelation with Target:")

print(target_correlation)

# ============================================================
# DATA PREPROCESSING
# ============================================================

# ------------------------------------------------------------
# 19. DEFINE FEATURES AND TARGET
# ------------------------------------------------------------

X = df.drop(
    columns=["target"]
)

y = df["target"]

print("\nFeatures:")
display(X.head())

print("\nTarget:")
display(y.head())

# ------------------------------------------------------------
# 20. CHECK UNIQUE TARGET CLASSES
# ------------------------------------------------------------

print("\nTarget Classes:")
print(sorted(y.unique()))

print("\nNumber of Classes:")
print(y.nunique())

# ------------------------------------------------------------
# 21. TRAIN-TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Shape:")
print(X_train.shape)

print("\nTesting Shape:")
print(X_test.shape)

# ============================================================
# DECISION TREE - ENTROPY
# ============================================================

# ------------------------------------------------------------
# 22. CREATE ENTROPY TREE
# ------------------------------------------------------------

tree_entropy = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

# ------------------------------------------------------------
# 23. TRAIN ENTROPY TREE
# ------------------------------------------------------------

tree_entropy.fit(
    X_train,
    y_train
)

print(
    "\nDecision Tree using Entropy trained successfully."
)

# ------------------------------------------------------------
# 24. ENTROPY PREDICTION
# ------------------------------------------------------------

y_pred_entropy = tree_entropy.predict(
    X_test
)

# ------------------------------------------------------------
# 25. ENTROPY ACCURACY
# ------------------------------------------------------------

entropy_accuracy = accuracy_score(
    y_test,
    y_pred_entropy
)

print("\nEntropy Tree Accuracy:")
print(entropy_accuracy)

# ============================================================
# DECISION TREE - GINI
# ============================================================

# ------------------------------------------------------------
# 26. CREATE GINI TREE
# ------------------------------------------------------------

tree_gini = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)

# ------------------------------------------------------------
# 27. TRAIN GINI TREE
# ------------------------------------------------------------

tree_gini.fit(
    X_train,
    y_train
)

print(
    "\nDecision Tree using Gini trained successfully."
)

# ------------------------------------------------------------
# 28. GINI PREDICTION
# ------------------------------------------------------------

y_pred_gini = tree_gini.predict(
    X_test
)

# ------------------------------------------------------------
# 29. GINI ACCURACY
# ------------------------------------------------------------

gini_accuracy = accuracy_score(
    y_test,
    y_pred_gini
)

print("\nGini Tree Accuracy:")
print(gini_accuracy)

# ============================================================
# ENTROPY MODEL EVALUATION
# ============================================================

# ------------------------------------------------------------
# 30. PRECISION
# ------------------------------------------------------------

entropy_precision = precision_score(
    y_test,
    y_pred_entropy,
    average="weighted",
    zero_division=0
)

# ------------------------------------------------------------
# 31. RECALL
# ------------------------------------------------------------

entropy_recall = recall_score(
    y_test,
    y_pred_entropy,
    average="weighted",
    zero_division=0
)

# ------------------------------------------------------------
# 32. F1 SCORE
# ------------------------------------------------------------

entropy_f1 = f1_score(
    y_test,
    y_pred_entropy,
    average="weighted",
    zero_division=0
)

print("\n==============================")
print("ENTROPY TREE RESULTS")
print("==============================")

print(
    f"Accuracy : {entropy_accuracy:.4f}"
)

print(
    f"Precision: {entropy_precision:.4f}"
)

print(
    f"Recall   : {entropy_recall:.4f}"
)

print(
    f"F1 Score : {entropy_f1:.4f}"
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred_entropy,
        zero_division=0
    )
)

# ============================================================
# GINI MODEL EVALUATION
# ============================================================

# ------------------------------------------------------------
# 33. PRECISION
# ------------------------------------------------------------

gini_precision = precision_score(
    y_test,
    y_pred_gini,
    average="weighted",
    zero_division=0
)

# ------------------------------------------------------------
# 34. RECALL
# ------------------------------------------------------------

gini_recall = recall_score(
    y_test,
    y_pred_gini,
    average="weighted",
    zero_division=0
)

# ------------------------------------------------------------
# 35. F1 SCORE
# ------------------------------------------------------------

gini_f1 = f1_score(
    y_test,
    y_pred_gini,
    average="weighted",
    zero_division=0
)

print("\n==============================")
print("GINI TREE RESULTS")
print("==============================")

print(
    f"Accuracy : {gini_accuracy:.4f}"
)

print(
    f"Precision: {gini_precision:.4f}"
)

print(
    f"Recall   : {gini_recall:.4f}"
)

print(
    f"F1 Score : {gini_f1:.4f}"
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred_gini,
        zero_division=0
    )
)

# ============================================================
# CONFUSION MATRIX
# ============================================================

# ------------------------------------------------------------
# 36. ENTROPY CONFUSION MATRIX
# ------------------------------------------------------------

cm_entropy = confusion_matrix(
    y_test,
    y_pred_entropy
)

print("\nEntropy Confusion Matrix:")
print(cm_entropy)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_entropy,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Decision Tree - Entropy Confusion Matrix"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()

# ------------------------------------------------------------
# 37. GINI CONFUSION MATRIX
# ------------------------------------------------------------

cm_gini = confusion_matrix(
    y_test,
    y_pred_gini
)

print("\nGini Confusion Matrix:")
print(cm_gini)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_gini,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Decision Tree - Gini Confusion Matrix"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()

# ============================================================
# TREE INFORMATION
# ============================================================

# ------------------------------------------------------------
# 38. TREE DEPTH
# ------------------------------------------------------------

print("\nEntropy Tree Depth:")
print(tree_entropy.get_depth())

print("\nEntropy Tree Number of Leaves:")
print(tree_entropy.get_n_leaves())

print("\nGini Tree Depth:")
print(tree_gini.get_depth())

print("\nGini Tree Number of Leaves:")
print(tree_gini.get_n_leaves())

# ============================================================
# FEATURE IMPORTANCE
# ============================================================

# ------------------------------------------------------------
# 39. ENTROPY FEATURE IMPORTANCE
# ------------------------------------------------------------

entropy_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": tree_entropy.feature_importances_
})

entropy_importance = entropy_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nEntropy Feature Importance:")

display(entropy_importance)

# ------------------------------------------------------------
# 40. GINI FEATURE IMPORTANCE
# ------------------------------------------------------------

gini_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": tree_gini.feature_importances_
})

gini_importance = gini_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nGini Feature Importance:")

display(gini_importance)

# ------------------------------------------------------------
# 41. FEATURE IMPORTANCE PLOT
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.barh(
    entropy_importance["Feature"],
    entropy_importance["Importance"]
)

plt.title(
    "Decision Tree Feature Importance"
)

plt.xlabel("Importance")
plt.ylabel("Feature")

plt.gca().invert_yaxis()

plt.show()

# ============================================================
# DECISION TREE VISUALIZATION
# ============================================================

# ------------------------------------------------------------
# 42. VISUALIZE ENTROPY TREE
# ------------------------------------------------------------

plt.figure(figsize=(25, 15))

plot_tree(
    tree_entropy,
    feature_names=X.columns,
    class_names=[
        str(class_name)
        for class_name in tree_entropy.classes_
    ],
    filled=True,
    rounded=True,
    fontsize=8
)

plt.title(
    "Decision Tree Using Entropy"
)

plt.show()

# ============================================================
# DEPTH SWEEP
# ============================================================

# ------------------------------------------------------------
# 43. TEST DIFFERENT TREE DEPTHS
# ------------------------------------------------------------

depths = range(1, 16)

train_scores = []
test_scores = []

for depth in depths:

    model = DecisionTreeClassifier(
        criterion="entropy",
        max_depth=depth,
        random_state=42
    )

    model.fit(
        X_train,
        y_train
    )

    train_prediction = model.predict(
        X_train
    )

    test_prediction = model.predict(
        X_test
    )

    train_score = accuracy_score(
        y_train,
        train_prediction
    )

    test_score = accuracy_score(
        y_test,
        test_prediction
    )

    train_scores.append(
        train_score
    )

    test_scores.append(
        test_score
    )

# ------------------------------------------------------------
# 44. DEPTH RESULTS
# ------------------------------------------------------------

depth_results = pd.DataFrame({
    "Max Depth": list(depths),
    "Training Accuracy": train_scores,
    "Testing Accuracy": test_scores
})

print("\nDepth Sweep Results:")

display(depth_results)

# ------------------------------------------------------------
# 45. DEPTH VS ACCURACY
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.plot(
    depths,
    train_scores,
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    depths,
    test_scores,
    marker="o",
    label="Testing Accuracy"
)

plt.title(
    "Decision Tree Depth vs Accuracy"
)

plt.xlabel("Maximum Tree Depth")
plt.ylabel("Accuracy")

plt.legend()

plt.show()

# ============================================================
# COST-COMPLEXITY PRUNING
# ============================================================

# ------------------------------------------------------------
# 46. GET CCP ALPHAS
# ------------------------------------------------------------

path = tree_entropy.cost_complexity_pruning_path(
    X_train,
    y_train
)

ccp_alphas = path.ccp_alphas

print("\nNumber of CCP Alpha Values:")
print(len(ccp_alphas))

# ------------------------------------------------------------
# 47. TRAIN PRUNED TREES
# ------------------------------------------------------------

pruned_models = []

for alpha in ccp_alphas:

    model = DecisionTreeClassifier(
        criterion="entropy",
        random_state=42,
        ccp_alpha=alpha
    )

    model.fit(
        X_train,
        y_train
    )

    pruned_models.append(
        model
    )

# ------------------------------------------------------------
# 48. CALCULATE ACCURACY
# ------------------------------------------------------------

train_accuracy_pruned = []
test_accuracy_pruned = []

for model in pruned_models:

    train_prediction = model.predict(
        X_train
    )

    test_prediction = model.predict(
        X_test
    )

    train_accuracy_pruned.append(
        accuracy_score(
            y_train,
            train_prediction
        )
    )

    test_accuracy_pruned.append(
        accuracy_score(
            y_test,
            test_prediction
        )
    )

# ------------------------------------------------------------
# 49. PRUNING RESULTS
# ------------------------------------------------------------

pruning_results = pd.DataFrame({
    "CCP Alpha": ccp_alphas,
    "Training Accuracy": train_accuracy_pruned,
    "Testing Accuracy": test_accuracy_pruned
})

print("\nCost-Complexity Pruning Results:")

display(
    pruning_results.head(20)
)

# ------------------------------------------------------------
# 50. PRUNING PLOT
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.plot(
    ccp_alphas,
    train_accuracy_pruned,
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    ccp_alphas,
    test_accuracy_pruned,
    marker="o",
    label="Testing Accuracy"
)

plt.xlabel("CCP Alpha")
plt.ylabel("Accuracy")

plt.title(
    "Cost-Complexity Pruning"
)

plt.legend()

plt.show()

# ------------------------------------------------------------
# 51. FIND BEST PRUNED MODEL
# ------------------------------------------------------------

best_index = np.argmax(
    test_accuracy_pruned
)

best_alpha = ccp_alphas[
    best_index
]

best_pruned_model = pruned_models[
    best_index
]

best_pruned_accuracy = test_accuracy_pruned[
    best_index
]

print("\nBest CCP Alpha:")
print(best_alpha)

print("\nBest Pruned Tree Accuracy:")
print(best_pruned_accuracy)

print("\nBest Pruned Tree Depth:")
print(
    best_pruned_model.get_depth()
)

print("\nBest Pruned Tree Leaves:")
print(
    best_pruned_model.get_n_leaves()
)

# ============================================================
# FINAL MODEL EVALUATION
# ============================================================

# ------------------------------------------------------------
# 52. FINAL PRUNED PREDICTIONS
# ------------------------------------------------------------

y_pred_pruned = best_pruned_model.predict(
    X_test
)

# ------------------------------------------------------------
# 53. FINAL METRICS
# ------------------------------------------------------------

final_accuracy = accuracy_score(
    y_test,
    y_pred_pruned
)

final_precision = precision_score(
    y_test,
    y_pred_pruned,
    average="weighted",
    zero_division=0
)

final_recall = recall_score(
    y_test,
    y_pred_pruned,
    average="weighted",
    zero_division=0
)

final_f1 = f1_score(
    y_test,
    y_pred_pruned,
    average="weighted",
    zero_division=0
)

print("\n==============================")
print("FINAL PRUNED TREE RESULTS")
print("==============================")

print(
    f"Accuracy : {final_accuracy:.4f}"
)

print(
    f"Precision: {final_precision:.4f}"
)

print(
    f"Recall   : {final_recall:.4f}"
)

print(
    f"F1 Score : {final_f1:.4f}"
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred_pruned,
        zero_division=0
    )
)

# ------------------------------------------------------------
# 54. FINAL CONFUSION MATRIX
# ------------------------------------------------------------

final_cm = confusion_matrix(
    y_test,
    y_pred_pruned
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    final_cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Final Pruned Decision Tree Confusion Matrix"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()

# ============================================================
# MODEL COMPARISON
# ============================================================

# ------------------------------------------------------------
# 55. COMPARE GINI, ENTROPY AND PRUNED TREE
# ------------------------------------------------------------

comparison = pd.DataFrame({

    "Model": [
        "Decision Tree - Gini",
        "Decision Tree - Entropy",
        "Pruned Decision Tree"
    ],

    "Accuracy": [
        gini_accuracy,
        entropy_accuracy,
        final_accuracy
    ],

    "Precision": [
        gini_precision,
        entropy_precision,
        final_precision
    ],

    "Recall": [
        gini_recall,
        entropy_recall,
        final_recall
    ],

    "F1 Score": [
        gini_f1,
        entropy_f1,
        final_f1
    ]
})

print("\n==============================")
print("MODEL COMPARISON")
print("==============================")

display(comparison)

# ------------------------------------------------------------
# 56. ACCURACY COMPARISON
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

plt.bar(
    comparison["Model"],
    comparison["Accuracy"]
)

plt.title(
    "Decision Tree Model Accuracy Comparison"
)

plt.xlabel("Model")
plt.ylabel("Accuracy")

plt.xticks(rotation=20)

plt.show()

# ============================================================
# FINAL CONCLUSION
# ============================================================

print("""
============================================================
CONCLUSION
============================================================

1. The Heart Disease dataset was loaded and explored.

2. Missing values and duplicate rows were checked.

3. EDA was performed on important health-related features.

4. The dataset was divided into training and testing sets.

5. A Decision Tree classifier was trained using Gini impurity.

6. Another Decision Tree was trained using Entropy.

7. Gini and Entropy models were compared.

8. Feature importance was analyzed.

9. The structure of the Decision Tree was visualized.

10. Different maximum tree depths were tested to study
    underfitting and overfitting.

11. Cost-Complexity Pruning was performed using CCP Alpha.

12. The best pruned tree was selected based on test accuracy.

13. Final performance was evaluated using:
    - Accuracy
    - Precision
    - Recall
    - F1 Score
    - Confusion Matrix

14. Gini, Entropy and Pruned Decision Trees were compared.

============================================================
""")
```
