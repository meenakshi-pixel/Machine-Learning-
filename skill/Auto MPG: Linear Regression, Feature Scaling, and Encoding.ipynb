```python
# ============================================================
# PROJECT 7: AUTO MPG
# LINEAR REGRESSION, FEATURE SCALING, AND ENCODING
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

from google.colab import files

uploaded = files.upload()

# Change the filename if necessary
df = pd.read_csv("auto-mpg.csv")

print("Dataset loaded successfully!")

# ------------------------------------------------------------
# 3. DISPLAY DATASET
# ------------------------------------------------------------

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

# ------------------------------------------------------------
# 4. DATASET SHAPE
# ------------------------------------------------------------

print("\nDataset Shape:")
print(df.shape)

# ------------------------------------------------------------
# 5. COLUMN NAMES
# ------------------------------------------------------------

print("\nColumn Names:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 6. DATASET INFORMATION
# ------------------------------------------------------------

print("\nDataset Information:")
df.info()

# ------------------------------------------------------------
# 7. STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\nStatistical Summary:")
display(df.describe())

# ------------------------------------------------------------
# 8. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 9. CHECK DUPLICATES
# ------------------------------------------------------------

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 10. REMOVE DUPLICATES
# ------------------------------------------------------------

df = df.drop_duplicates()

print("\nShape After Removing Duplicates:")
print(df.shape)

# ============================================================
# DATA CLEANING
# ============================================================

# ------------------------------------------------------------
# 11. CLEAN COLUMN NAMES
# ------------------------------------------------------------

df.columns = df.columns.str.strip()

print("\nColumn Names:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 12. CHECK HORSEPOWER DATA TYPE
# ------------------------------------------------------------

print("\nHorsepower Data Type:")
print(df["horsepower"].dtype)

print("\nUnique Horsepower Values:")
print(df["horsepower"].unique()[:20])

# ------------------------------------------------------------
# 13. CONVERT HORSEPOWER TO NUMERIC
# ------------------------------------------------------------

df["horsepower"] = pd.to_numeric(
    df["horsepower"],
    errors="coerce"
)

# ------------------------------------------------------------
# 14. CHECK MISSING VALUES AGAIN
# ------------------------------------------------------------

print("\nMissing Values After Conversion:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 15. HANDLE MISSING HORSEPOWER
# ------------------------------------------------------------

df["horsepower"] = df["horsepower"].fillna(
    df["horsepower"].median()
)

print("\nMissing Values After Filling:")
print(df.isnull().sum())

# ============================================================
# EXPLORATORY DATA ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 16. MPG DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="mpg",
    bins=30,
    kde=True
)

plt.title("MPG Distribution")
plt.xlabel("Miles Per Gallon")
plt.ylabel("Frequency")

plt.show()

# ------------------------------------------------------------
# 17. HORSEPOWER DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="horsepower",
    bins=30,
    kde=True
)

plt.title("Horsepower Distribution")
plt.xlabel("Horsepower")
plt.ylabel("Frequency")

plt.show()

# ------------------------------------------------------------
# 18. WEIGHT DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="weight",
    bins=30,
    kde=True
)

plt.title("Vehicle Weight Distribution")
plt.xlabel("Weight")
plt.ylabel("Frequency")

plt.show()

# ------------------------------------------------------------
# 19. MPG VS HORSEPOWER
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="horsepower",
    y="mpg"
)

plt.title("MPG vs Horsepower")
plt.xlabel("Horsepower")
plt.ylabel("MPG")

plt.show()

# ------------------------------------------------------------
# 20. MPG VS WEIGHT
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="weight",
    y="mpg"
)

plt.title("MPG vs Vehicle Weight")
plt.xlabel("Weight")
plt.ylabel("MPG")

plt.show()

# ------------------------------------------------------------
# 21. MPG VS DISPLACEMENT
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="displacement",
    y="mpg"
)

plt.title("MPG vs Displacement")
plt.xlabel("Displacement")
plt.ylabel("MPG")

plt.show()

# ------------------------------------------------------------
# 22. MPG VS CYLINDERS
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="cylinders",
    y="mpg"
)

plt.title("MPG by Number of Cylinders")
plt.xlabel("Cylinders")
plt.ylabel("MPG")

plt.show()

# ------------------------------------------------------------
# 23. MODEL YEAR VS MPG
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="model_year",
    y="mpg"
)

plt.title("Model Year vs MPG")
plt.xlabel("Model Year")
plt.ylabel("MPG")

plt.show()

# ------------------------------------------------------------
# 24. ORIGIN DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="origin"
)

plt.title("Vehicle Origin Distribution")
plt.xlabel("Origin")
plt.ylabel("Count")

plt.show()

# ------------------------------------------------------------
# 25. CORRELATION MATRIX
# ------------------------------------------------------------

plt.figure(figsize=(10, 8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()

# ------------------------------------------------------------
# 26. CORRELATION WITH MPG
# ------------------------------------------------------------

correlation_with_mpg = (
    df.corr(numeric_only=True)["mpg"]
    .sort_values(ascending=False)
)

print("\nCorrelation with MPG:")
print(correlation_with_mpg)

# ============================================================
# FEATURE ENGINEERING
# ============================================================

# ------------------------------------------------------------
# 27. CREATE HORSEPOWER PER CYLINDER
# ------------------------------------------------------------

df["horsepower_per_cylinder"] = (
    df["horsepower"] /
    df["cylinders"]
)

# ------------------------------------------------------------
# 28. CREATE WEIGHT PER CYLINDER
# ------------------------------------------------------------

df["weight_per_cylinder"] = (
    df["weight"] /
    df["cylinders"]
)

print("\nFeature Engineered Dataset:")
display(
    df[
        [
            "horsepower",
            "cylinders",
            "horsepower_per_cylinder",
            "weight",
            "weight_per_cylinder"
        ]
    ].head()
)

# ============================================================
# ENCODING
# ============================================================

# ------------------------------------------------------------
# 29. CHECK ORIGIN VALUES
# ------------------------------------------------------------

print("\nOrigin Values:")
print(df["origin"].unique())

# ------------------------------------------------------------
# 30. ONE-HOT ENCODE ORIGIN
# ------------------------------------------------------------

df = pd.get_dummies(
    df,
    columns=["origin"],
    prefix="origin",
    drop_first=True,
    dtype=int
)

print("\nDataset After Encoding:")
display(df.head())

print("\nColumns After Encoding:")
print(df.columns.tolist())

# ============================================================
# MACHINE LEARNING
# ============================================================

# ------------------------------------------------------------
# 31. DEFINE FEATURES AND TARGET
# ------------------------------------------------------------

X = df.drop(
    columns=[
        "mpg",
        "car_name"
    ],
    errors="ignore"
)

y = df["mpg"]

print("\nFeatures:")
display(X.head())

print("\nTarget:")
display(y.head())

# ------------------------------------------------------------
# 32. CHECK DATA TYPES
# ------------------------------------------------------------

print("\nFeature Data Types:")
print(X.dtypes)

# ------------------------------------------------------------
# 33. TRAIN-TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Shape:")
print(X_train.shape)

print("\nTesting Shape:")
print(X_test.shape)

# ------------------------------------------------------------
# 34. FEATURE SCALING
# ------------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)

print("\nFeature scaling completed.")

# ============================================================
# LINEAR REGRESSION
# ============================================================

# ------------------------------------------------------------
# 35. CREATE MODEL
# ------------------------------------------------------------

model = LinearRegression()

# ------------------------------------------------------------
# 36. TRAIN MODEL
# ------------------------------------------------------------

model.fit(
    X_train_scaled,
    y_train
)

print("\nLinear Regression model trained successfully.")

# ------------------------------------------------------------
# 37. PREDICTIONS
# ------------------------------------------------------------

y_pred = model.predict(
    X_test_scaled
)

# ============================================================
# MODEL EVALUATION
# ============================================================

# ------------------------------------------------------------
# 38. MAE
# ------------------------------------------------------------

mae = mean_absolute_error(
    y_test,
    y_pred
)

# ------------------------------------------------------------
# 39. MSE
# ------------------------------------------------------------

mse = mean_squared_error(
    y_test,
    y_pred
)

# ------------------------------------------------------------
# 40. RMSE
# ------------------------------------------------------------

rmse = np.sqrt(mse)

# ------------------------------------------------------------
# 41. R2 SCORE
# ------------------------------------------------------------

r2 = r2_score(
    y_test,
    y_pred
)

print("\n==============================")
print("LINEAR REGRESSION RESULTS")
print("==============================")

print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")

# ============================================================
# ACTUAL VS PREDICTED
# ============================================================

# ------------------------------------------------------------
# 42. ACTUAL VS PREDICTED PLOT
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.6
)

plt.xlabel("Actual MPG")
plt.ylabel("Predicted MPG")

plt.title(
    "Actual vs Predicted MPG"
)

plt.show()

# ------------------------------------------------------------
# 43. RESIDUALS
# ------------------------------------------------------------

residuals = (
    y_test -
    y_pred
)

plt.figure(figsize=(8, 5))

plt.scatter(
    y_pred,
    residuals,
    alpha=0.6
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title(
    "Residual Plot"
)

plt.xlabel("Predicted MPG")
plt.ylabel("Residual")

plt.show()

# ============================================================
# MODEL COEFFICIENTS
# ============================================================

# ------------------------------------------------------------
# 44. FEATURE COEFFICIENTS
# ------------------------------------------------------------

coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

coefficients["Absolute_Coefficient"] = (
    coefficients["Coefficient"].abs()
)

coefficients = coefficients.sort_values(
    by="Absolute_Coefficient",
    ascending=False
)

print("\nFeature Coefficients:")

display(coefficients)

# ------------------------------------------------------------
# 45. COEFFICIENT VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(10, 7))

plt.barh(
    coefficients["Feature"],
    coefficients["Coefficient"]
)

plt.title(
    "Linear Regression Feature Coefficients"
)

plt.xlabel("Coefficient")
plt.ylabel("Feature")

plt.gca().invert_yaxis()

plt.show()

# ============================================================
# PREDICTION TABLE
# ============================================================

# ------------------------------------------------------------
# 46. ACTUAL VS PREDICTED TABLE
# ------------------------------------------------------------

prediction_results = pd.DataFrame({
    "Actual_MPG": y_test.values,
    "Predicted_MPG": y_pred
})

prediction_results["Error"] = (
    prediction_results["Actual_MPG"] -
    prediction_results["Predicted_MPG"]
)

print("\nPrediction Results:")

display(
    prediction_results.head(20)
)

# ============================================================
# SAMPLE VEHICLE PREDICTION
# ============================================================

# ------------------------------------------------------------
# 47. SELECT ONE SAMPLE FROM TEST DATA
# ------------------------------------------------------------

sample = X_test.iloc[[0]]

print("\nSample Vehicle Features:")
display(sample)

# ------------------------------------------------------------
# 48. SCALE SAMPLE
# ------------------------------------------------------------

sample_scaled = scaler.transform(
    sample
)

# ------------------------------------------------------------
# 49. PREDICT MPG
# ------------------------------------------------------------

sample_prediction = model.predict(
    sample_scaled
)[0]

print("\n==============================")
print("SAMPLE VEHICLE PREDICTION")
print("==============================")

print(
    f"Predicted MPG: {sample_prediction:.2f}"
)

# ============================================================
# MODEL SUMMARY
# ============================================================

# ------------------------------------------------------------
# 50. RESULTS TABLE
# ------------------------------------------------------------

results = pd.DataFrame({
    "Metric": [
        "MAE",
        "MSE",
        "RMSE",
        "R2 Score"
    ],

    "Value": [
        mae,
        mse,
        rmse,
        r2
    ]
})

print("\n==============================")
print("MODEL PERFORMANCE")
print("==============================")

display(results)

# ============================================================
# SAVE PROCESSED DATASET
# ============================================================

# ------------------------------------------------------------
# 51. SAVE DATASET
# ------------------------------------------------------------

df.to_csv(
    "Auto_MPG_Processed.csv",
    index=False
)

print(
    "\nProcessed dataset saved as "
    "Auto_MPG_Processed.csv"
)

# ------------------------------------------------------------
# 52. DOWNLOAD DATASET
# ------------------------------------------------------------

files.download(
    "Auto_MPG_Processed.csv"
)

# ============================================================
# FINAL CONCLUSION
# ============================================================

print("""
============================================================
CONCLUSION
============================================================

1. The Auto MPG dataset was loaded and explored.

2. Missing horsepower values were converted to numeric
   and handled using the median.

3. Duplicate records were checked and removed.

4. EDA was performed on MPG, horsepower, weight,
   displacement, cylinders and model year.

5. New features were created:
   - horsepower_per_cylinder
   - weight_per_cylinder

6. The categorical Origin feature was one-hot encoded.

7. The dataset was divided into training and testing sets.

8. Numerical features were standardized using
   StandardScaler.

9. Linear Regression was trained to predict MPG.

10. Model performance was evaluated using:
    - MAE
    - MSE
    - RMSE
    - R² Score

11. Actual and predicted MPG values were compared.

12. Feature coefficients were analyzed to understand
    their influence on MPG.

============================================================
""")
```
