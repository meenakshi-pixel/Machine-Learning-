```python
# ============================================================
# PROJECT 3: ADULT INCOME
# FEATURE ENGINEERING AND EDA
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

# Upload the Adult Income CSV file
from google.colab import files

uploaded = files.upload()

# Replace the filename if your uploaded file has a different name
df = pd.read_csv("adult.csv")

print("Dataset loaded successfully!")

# ------------------------------------------------------------
# 3. DISPLAY DATA
# ------------------------------------------------------------

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

# ------------------------------------------------------------
# 4. DATASET SHAPE
# ------------------------------------------------------------

print("\nDataset Shape:")
print(df.shape)

# ------------------------------------------------------------
# 5. COLUMN NAMES
# ------------------------------------------------------------

print("\nColumn Names:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 6. DATASET INFORMATION
# ------------------------------------------------------------

print("\nDataset Information:")
df.info()

# ------------------------------------------------------------
# 7. STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\nNumerical Statistical Summary:")
display(df.describe())

print("\nCategorical Statistical Summary:")
display(df.describe(include="object"))

# ------------------------------------------------------------
# 8. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 9. CHECK DUPLICATES
# ------------------------------------------------------------

print("\nDuplicate Rows:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 10. REMOVE DUPLICATES
# ------------------------------------------------------------

df = df.drop_duplicates()

print("\nShape After Removing Duplicates:")
print(df.shape)

# ------------------------------------------------------------
# 11. CLEAN COLUMN NAMES
# ------------------------------------------------------------

df.columns = df.columns.str.strip()

print("\nCleaned Column Names:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 12. CLEAN STRING COLUMNS
# ------------------------------------------------------------

for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].str.strip()

# ------------------------------------------------------------
# 13. CHECK UNIQUE VALUES
# ------------------------------------------------------------

print("\nUnique Values in Categorical Columns:")

for column in df.select_dtypes(include="object").columns:
    print("\n", column)
    print(df[column].unique()[:20])

# ------------------------------------------------------------
# 14. INCOME DISTRIBUTION
# ------------------------------------------------------------

print("\nIncome Distribution:")
print(df["income"].value_counts())

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="income"
)

plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Count")
plt.show()

# ------------------------------------------------------------
# 15. INCOME PERCENTAGE
# ------------------------------------------------------------

income_percentage = (
    df["income"]
    .value_counts(normalize=True)
    * 100
)

print("\nIncome Percentage:")
print(income_percentage)

# ------------------------------------------------------------
# 16. AGE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="age",
    bins=30,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# ------------------------------------------------------------
# 17. AGE VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="income",
    y="age"
)

plt.title("Age Distribution by Income")
plt.xlabel("Income")
plt.ylabel("Age")
plt.show()

# ------------------------------------------------------------
# 18. EDUCATION DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    y="education",
    order=df["education"].value_counts().index
)

plt.title("Education Distribution")
plt.xlabel("Count")
plt.ylabel("Education")
plt.show()

# ------------------------------------------------------------
# 19. EDUCATION VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    y="education",
    hue="income"
)

plt.title("Education Level vs Income")
plt.xlabel("Count")
plt.ylabel("Education")
plt.show()

# ------------------------------------------------------------
# 20. HOURS PER WEEK
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="hours-per-week",
    bins=30,
    kde=True
)

plt.title("Working Hours per Week")
plt.xlabel("Hours per Week")
plt.ylabel("Frequency")
plt.show()

# ------------------------------------------------------------
# 21. HOURS PER WEEK VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="income",
    y="hours-per-week"
)

plt.title("Working Hours vs Income")
plt.xlabel("Income")
plt.ylabel("Hours per Week")
plt.show()

# ------------------------------------------------------------
# 22. GENDER DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="sex"
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()

# ------------------------------------------------------------
# 23. GENDER VS INCOME
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="sex",
    hue="income"
)

plt.title("Gender vs Income")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()

# ------------------------------------------------------------
# 24. WORKCLASS DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    y="workclass",
    order=df["workclass"].value_counts().index
)

plt.title("Workclass Distribution")
plt.xlabel("Count")
plt.ylabel("Workclass")
plt.show()

# ------------------------------------------------------------
# 25. OCCUPATION DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(12, 7))

sns.countplot(
    data=df,
    y="occupation",
    order=df["occupation"].value_counts().index
)

plt.title("Occupation Distribution")
plt.xlabel("Count")
plt.ylabel("Occupation")
plt.show()

# ------------------------------------------------------------
# 26. CAPITAL GAIN DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="capital-gain",
    bins=30
)

plt.title("Capital Gain Distribution")
plt.xlabel("Capital Gain")
plt.ylabel("Frequency")
plt.show()

# ------------------------------------------------------------
# 27. CAPITAL LOSS DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="capital-loss",
    bins=30
)

plt.title("Capital Loss Distribution")
plt.xlabel("Capital Loss")
plt.ylabel("Frequency")
plt.show()

# ============================================================
# FEATURE ENGINEERING
# ============================================================

# ------------------------------------------------------------
# 28. CREATE AGE GROUP
# ------------------------------------------------------------

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 25, 35, 50, 65, 100],
    labels=[
        "Young",
        "Adult",
        "Middle_Aged",
        "Senior",
        "Elderly"
    ]
)

print("\nAge Group:")
display(df[["age", "age_group"]].head(10))

# ------------------------------------------------------------
# 29. CREATE CAPITAL NET
# ------------------------------------------------------------

df["capital_net"] = (
    df["capital-gain"] -
    df["capital-loss"]
)

print("\nCapital Net:")
display(
    df[
        [
            "capital-gain",
            "capital-loss",
            "capital_net"
        ]
    ].head()
)

# ------------------------------------------------------------
# 30. CREATE WORKING HOURS CATEGORY
# ------------------------------------------------------------

df["work_hours_category"] = pd.cut(
    df["hours-per-week"],
    bins=[0, 30, 40, 60, 100],
    labels=[
        "Part-Time",
        "Full-Time",
        "Overtime",
        "Extreme"
    ]
)

print("\nWorking Hours Category:")
display(
    df[
        [
            "hours-per-week",
            "work_hours_category"
        ]
    ].head(10)
)

# ------------------------------------------------------------
# 31. CREATE EDUCATION NUMBER
# ------------------------------------------------------------

education_mapping = {
    "Preschool": 0,
    "1st-4th": 1,
    "5th-6th": 2,
    "7th-8th": 3,
    "9th": 4,
    "10th": 5,
    "11th": 6,
    "12th": 7,
    "HS-grad": 8,
    "Some-college": 9,
    "Assoc-voc": 10,
    "Assoc-acdm": 11,
    "Bachelors": 12,
    "Masters": 13,
    "Prof-school": 14,
    "Doctorate": 15
}

df["education_num_custom"] = (
    df["education"]
    .map(education_mapping)
)

print("\nEducation Mapping:")
display(
    df[
        [
            "education",
            "education_num_custom"
        ]
    ].head(15)
)

# ------------------------------------------------------------
# 32. INCOME BINARY FEATURE
# ------------------------------------------------------------

df["income_binary"] = df["income"].map({
    "<=50K": 0,
    ">50K": 1,
    "<=50K.": 0,
    ">50K.": 1
})

print("\nBinary Income:")
print(df["income_binary"].value_counts())

# ------------------------------------------------------------
# 33. INCOME BY AGE GROUP
# ------------------------------------------------------------

age_income = pd.crosstab(
    df["age_group"],
    df["income"],
    normalize="index"
) * 100

print("\nIncome Percentage by Age Group:")
display(age_income)

# ------------------------------------------------------------
# 34. INCOME BY EDUCATION
# ------------------------------------------------------------

education_income = pd.crosstab(
    df["education"],
    df["income"],
    normalize="index"
) * 100

print("\nIncome Percentage by Education:")
display(education_income)

# ------------------------------------------------------------
# 35. INCOME BY GENDER
# ------------------------------------------------------------

gender_income = pd.crosstab(
    df["sex"],
    df["income"],
    normalize="index"
) * 100

print("\nIncome Percentage by Gender:")
display(gender_income)

# ------------------------------------------------------------
# 36. INCOME BY WORK HOURS CATEGORY
# ------------------------------------------------------------

hours_income = pd.crosstab(
    df["work_hours_category"],
    df["income"],
    normalize="index"
) * 100

print("\nIncome Percentage by Working Hours:")
display(hours_income)

# ------------------------------------------------------------
# 37. NUMERICAL CORRELATION
# ------------------------------------------------------------

numeric_columns = df.select_dtypes(
    include=np.number
)

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_columns.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Numerical Feature Correlation")
plt.show()

# ------------------------------------------------------------
# 38. GROUP ANALYSIS
# ------------------------------------------------------------

print("\nAverage Age by Income:")
print(
    df.groupby("income")["age"].mean()
)

print("\nAverage Hours per Week by Income:")
print(
    df.groupby("income")["hours-per-week"].mean()
)

print("\nAverage Capital Gain by Income:")
print(
    df.groupby("income")["capital-gain"].mean()
)

# ------------------------------------------------------------
# 39. FINAL DATASET INFORMATION
# ------------------------------------------------------------

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nFinal Columns:")
print(df.columns.tolist())

print("\nFinal Missing Values:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 40. SAVE FEATURE-ENGINEERED DATASET
# ------------------------------------------------------------

df.to_csv(
    "Adult_Income_Feature_Engineered.csv",
    index=False
)

print(
    "\nDataset saved as "
    "Adult_Income_Feature_Engineered.csv"
)

# ------------------------------------------------------------
# 41. DOWNLOAD DATASET
# ------------------------------------------------------------

files.download(
    "Adult_Income_Feature_Engineered.csv"
)
```
